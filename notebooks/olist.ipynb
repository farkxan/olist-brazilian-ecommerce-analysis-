```python
orders['order_month'] = (
    orders['order_purchase_timestamp']
    .dt.to_period('M')
)
```

```python
monthly_orders = (
    orders
    .groupby('order_month')
    .size()
)

## SELECT
DATE_TRUNC('month', order_purchase_timestamp),
COUNT(*)
FROM orders
GROUP BY 1;
```

```python
monthly_orders.head()
```

```python
monthly_orders = (
    orders
    .groupby('order_month')
    .size()
    .reset_index(name='total_orders')
)
```

```python
monthly_orders.head() ## kalo di sql: SELECT
    ##order_month,
    ##COUNT(*) AS total_orders
##FROM orders
##GROUP BY order_month;
```

## Business question:
"Apakah jumlah order meningkat dibanding bulan sebelumnya?"

```python
## buat kolom baru
monthly_orders['previous_orders'] = (
    monthly_orders['total_orders']
    .shift(1)
)
```

```python
monthly_orders['growth'] = (
    (monthly_orders['total_orders'] -
     monthly_orders['previous_orders'])
    /
    monthly_orders['previous_orders']
    *100
)
```
